# This Notebook Processes Example Questions, Answers and Judgements of selected Models

```python
import pandas as pd
import numpy as np
import re
import json
import os

from itertools import islice
```

```python
bench_name = 'arena-hard-v0.1'

judge_model = 'gpt-4o'
baseline_model = 'gpt-4-0314'

model_list = ['gpt-4o-mini', 'mistral_large_2', 'mistral_nemo', 'mistral_nemo_fp8', 'mistral_nemo_int4', 'llama3_1_405b', 'llama3_1_70b', 'llama3_1_70b_fp8_dyn', 'llama3_1_70b_awq_int4', 'llama3_1_8b', 'llama3_1_8b_fp8', 'llama3_1_8b_int4', 'llama3_70b']
guided_model_list = [model_name + '_guided' for model_name in model_list]

question_file = f"arena-hard-auto/data/{bench_name}/question.jsonl"
guidance_file = f"arena-hard-auto/data/{bench_name}/guidance.jsonl"


if os.path.exists(question_file):
    print(f"Found question file at '{question_file}'")

if os.path.exists(guidance_file):
    print(f"Found guidance file at '{guidance_file}'")
```

```python
def get_questions(question_file: str):
    """Load questions from a file into a list."""

    questions = {}
    with open(question_file, "r", encoding='utf-8') as ques_file:
        for line in ques_file:
            if line:
                line = json.loads(line)
                questions[line["question_id"]] = line["turns"][0]["content"]
    return questions
```

```python
def get_ques_clusters(question_file: str):
    """Load questions from a file into a list."""

    clusters = {}
    with open(question_file, "r", encoding='utf-8') as ques_file:
        for line in ques_file:
            if line:
                line = json.loads(line)
                clusters[line["question_id"]] = line["cluster"]
    return clusters
```

```python
def get_answers(answer_file: str):
    """Load model answers."""

    answers = {}
    with open(answer_file, "r", encoding='utf-8') as ans_file:
        for line in ans_file:
            if line:
                line = json.loads(line)
                answers[line["question_id"]] = line["choices"][0]["turns"][0]["content"]
    return answers
```

```python
def get_judgments(judgment_file: str):
    """Load model judgements."""


    judgments = {}
    with open(judgment_file, "r", encoding='utf-8') as judge_file:
        for line in judge_file:
            if line:
                line = json.loads(line)
                # In The first game Assistant A is the Baseline. In the 2nd Game Assistant B is the Baseline
                judgments[line["question_id"]] = line["games"][1]["judgment"]
    return judgments
```

```python
def get_score(judgment_file: str):
    """Load model judgements."""

    scores = {}
    with open(judgment_file, "r", encoding='utf-8') as judge_file:
        for line in judge_file:
            if line:
                line = json.loads(line)
                # In The first game Assistant A is the Baseline. In the 2nd Game Assistant B is the Baseline
                scores[line["question_id"]] = line["games"][1]["score"]
    return scores
```

```python
def load_guidance(guidance_file: str):
    """Load guidances from a file."""
    guidances = {}
    with open(guidance_file, "r", encoding='utf-8') as fin:
        for line in fin:
            if line:
                line = json.loads(line)
                guidances[line["question_id"]] = line["guidance"]
    return guidances
```

```python
all_questions = get_questions(question_file)
all_guidances = load_guidance(guidance_file)
all_clusters = get_ques_clusters(question_file)
baseline_answer = get_answers(f'arena-hard-auto/data/{bench_name}/model_answer/{baseline_model}.jsonl')
judge_answer = get_answers(f'arena-hard-auto/data/{bench_name}/model_answer/{judge_model}.jsonl')

all_answers = []
all_guided_answers = []

all_judgments = []
all_guided_judgments = []

all_scores = []
all_guided_scores = []

for model_name in model_list:
    answer_dir_name = f'arena-hard-auto/data/{bench_name}/model_answer/{model_name}.jsonl'
    if not os.path.exists(answer_dir_name):
        print(f"No Bench results file found at '{answer_dir_name}'")
    else: 
        #print(f"Found Bench results file at '{answer_dir_name}'")
        all_answers.append(get_answers(answer_dir_name))
    
    judge_dir_name = f'arena-hard-auto/data/{bench_name}/model_judgment/gpt-4o-batch_2/{model_name}.jsonl'
    if not os.path.exists(judge_dir_name):
        print(f"No Bench results file found at '{judge_dir_name}'")
    else:
        all_judgments.append(get_judgments(judge_dir_name))
        all_scores.append(get_score(judge_dir_name))

for model_name in guided_model_list:
    answer_dir_name = f'arena-hard-auto/data/{bench_name}/model_answer/{model_name}.jsonl'
    if not os.path.exists(answer_dir_name):
        print(f"No Bench results file found at '{answer_dir_name}'")
    else: 
        #print(f"Found Bench results file at '{answer_dir_name}'")
        all_guided_answers.append(get_answers(answer_dir_name))
    
    judge_dir_name = f'arena-hard-auto/data/{bench_name}/model_judgment/gpt-4o-batch_2/{model_name}.jsonl'
    if not os.path.exists(judge_dir_name):
        print(f"No Bench results file found at '{judge_dir_name}'")
    else:
        all_guided_judgments.append(get_judgments(judge_dir_name))
        all_guided_scores.append(get_score(judge_dir_name))
```

```python
data = []

for qid, question in selected_questions:
    system_prompt = f"""
            You are a sophisticated AI-Expert there to help users solve tasks in several domains efficiently and accurately.
            Now solve the following task from the domain "{selected_clusters[qid]}".\n
            """
    
    guided_system_prompt = f"""
            You are a sophisticated AI assistant with the task to solve a question in the domain of {selected_clusters[qid]}. 
            Here is important guidance to solve the task that will be given to you:\n
            """
    
    judge_prompt_start = f'''Please act as an impartial judge and evaluate the quality of the responses provided by two AI assistants to the user prompt.\n'''

    judge_prompt = f'''
            You will be given assistant A's answer and assistant B's answer. 
            Your job is to evaluate which assistant's answer is better.\n\n
            Begin your evaluation by generating your own answer to the prompt. 
            You must provide your answers before judging any answers.\n\n
            When evaluating the assistants' answers, compare both assistants' answers with your answer. 
            You must identify and correct any mistakes or inaccurate information.\n\n
            Then consider if the assistant's answers are helpful, relevant, and concise. 
            Helpful means the answer correctly responds to the prompt or follows the instructions. 
            Note when user prompt has any ambiguity or more than one interpretation, it is more helpful and appropriate to ask for clarifications or more information from the user than providing an answer based on assumptions. 
            Relevant means all parts of the response closely connect or are appropriate to what is being asked. 
            Concise means the response is clear and not verbose or excessive.\n\n
            Then consider the creativity and novelty of the assistant's answers when needed. 
            Finally, identify any missing important information in the assistants' answers that would be beneficial to include when responding to the user prompt.\n\n
            After providing your explanation, you must output only one of the following choices as your final verdict with a label:\n\n
            1. Assistant A is significantly better: [[A>>B]]\n
            2. Assistant A is slightly better: [[A>B]]\n
            3. Tie, relatively the same: [[A=B]]\n
            4. Assistant B is slightly better: [[B>A]]\n
            5. Assistant B is significantly better: [[B>>A]]\n\n
            Example output: \"My final verdict is tie: [[A=B]]\".
            '''
    
    question_formatted = f"""
            Your task is:\n
            {question}\n
            """

    row = {"qid": qid, "question": question_formatted, "cluster": selected_clusters[qid], "guidance": selected_quidances[qid], "system_prompt": system_prompt, "guided_system_prompt": guided_system_prompt, "judge_prompt_start": judge_prompt_start, "judge_prompt": judge_prompt, "judge_answer_gpt4o": judge_answer[qid] ,"baseline_answer_gpt4-0314": baseline_answer[qid]}
    for model_name in model_list:
        if model_name == 'llama3_1_70b_fp8_dyn':
            col_model_name = 'llama3_1_70b_fp8'
        elif model_name == 'llama3_1_70b_awq_int4':
            col_model_name = 'llama3_1_70b_int4'
        else:
            col_model_name = model_name

        row[col_model_name] = selected_answers[model_name][qid]
        row[col_model_name + "_judgment"] = selected_judgments[model_name][qid]
        row[col_model_name + "_score"] = selected_scores[model_name][qid]

    for model_name in guided_model_list:
        if model_name == 'llama3_1_70b_fp8_dyn_guided':
            col_model_name = 'llama3_1_70b_fp8_guided'
        elif model_name == 'llama3_1_70b_awq_int4_guided':
            col_model_name = 'llama3_1_70b_int4_guided'
        else:
            col_model_name = model_name
        
        row[col_model_name] = selected_guided_answers[model_name][qid]
        row[col_model_name + "_judgment"] = selected_guided_judgments[model_name][qid]
        row[col_model_name + "_score"] = selected_guided_scores[model_name][qid]
    data.append(row)

df = pd.DataFrame(data)


df
```
